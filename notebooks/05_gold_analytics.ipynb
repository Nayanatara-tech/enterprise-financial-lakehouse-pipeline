###Revenue by region

```python
%sql
CREATE OR REPLACE TABLE finance_dev.gold_revenue_by_region
AS
SELECT
Region,
round(sum(AmountUSD),2) as Revenue
from finance_dev.silver_transactions_enriched
group by Region
ORDER BY revenue desc;

select * from finance_dev.gold_revenue_by_region;
```

###Top Customers by region

```python
%sql
CREATE OR REPLACE TABLE finance_dev.gold_top_customers_by_region
AS
SELECT
Region,
CustomerName,
round(sum(AmountUSD),2) as Revenue
from finance_dev.silver_transactions_enriched
group by Region,CustomerName
ORDER BY revenue desc;

select * from finance_dev.gold_top_customers_by_region;
```

###Monthly Revenue Trend

```python
%sql
CREATE OR REPLACE TABLE finance_dev.gold_monthly_trend AS
SELECT
    date_format(to_date(TransactionDate), 'yyyy-MM') AS YearMonth,
    ROUND(SUM(AmountUSD), 2) AS RevenueUSD
FROM finance_dev.silver_transactions_enriched
GROUP BY YearMonth
ORDER BY YearMonth;

select * from finance_dev.gold_monthly_trend
```

###Currency Distribution percentage

```python
%sql
CREATE OR REPLACE TABLE finance_dev.gold_currency_distribution as
(With currency_amt as
(
SELECT
Currency,
round(sum(Amount),2) as Revenue
from finance_dev.silver_transactions_enriched
group by Currency
)

select
Currency,
Revenue,
round(Revenue/sum(Revenue) over(),2)*100 as distribution_currency
from currency_amt);

select * from finance_dev.gold_currency_distribution
```
